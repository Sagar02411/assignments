```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['John', 'Anna', 'Mike']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [5000, 6000, 7000]
})

merged = pd.merge(df1, df2, on='ID')

print(merged)
```

```python
import pandas as pd

data = {
    'Department': ['HR', 'IT', 'HR', 'IT'],
    'Employee': ['A', 'B', 'C', 'D'],
    'Salary': [3000, 4000, 3500, 4500]
}

df = pd.DataFrame(data)

pivot = df.pivot_table(
    values='Salary',
    index='Department',
    aggfunc='mean'
)

print(pivot)
```
